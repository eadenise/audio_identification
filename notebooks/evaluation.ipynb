# F1, Precision and Recall

```python
import glob
import os
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path
```

```python
# Load file
result_file = Path('../output.txt')

genre_stats = {}

with open(result_file, "r") as f:
    for line in f:

        parts = line.strip().split()

           
        if len(parts) >= 4:
            qname = parts[1]
            ground_truth = qname.split('-')[0] + ".wav"

            true_genre = qname.split('.')[0]
            
        
            db_predictions = parts[3:]
            first_song = db_predictions[0]
            pred_genre = first_song.split(".")[0]

            for g in [true_genre, pred_genre]:
                if g not in genre_stats:
                    genre_stats[g] = {"TP": 0,"FN": 0, "FP": 0, "total": 0}

            genre_stats[true_genre]["total"]+=1
        
            if ground_truth == first_song:
                genre_stats[true_genre]["TP"]+=1      
        
            else:
                genre_stats[true_genre]["FN"]+=1
                genre_stats[pred_genre]["FP"] += 1
               
print("PRECISION, RECALL AND F1 by Genre")

for g_name, stats in genre_stats.items():
    tp = stats["TP"]
    fn = stats["FN"]
    fp = stats["FP"]
    total = stats["total"]
    
    # Calculate metrics (avoiding division by zero)
    recall = tp / (tp + fn) if (tp + fn) > 0 else 0.0
    precision = tp / (tp + fp) if (tp + fp) > 0 else 0.0
    f1_score = (2 * precision * recall) / (precision + recall) if (precision + recall) > 0 else 0.0
    
    print(f"Genre: {g_name.upper()} (Total Queries: {total})")
    print(f"True Positives:  {tp}")
    print(f"False Negatives: {fn}")
    print(f"False Positives: {fp}")
    print(f"Precision: {precision*100:.1f}%")
    print(f"Recall:    {recall*100:.1f}%")
    print(f"F1 Score:  {f1_score*100:.1f}%\n")
```

```python
# Overall Metrics
all_precisions = []
all_recalls = []
all_f1s = []

for g_name, stats in genre_stats.items():
    tp = stats["TP"]
    fn = stats["FN"]
    fp = stats["FP"]
    
    p = tp / (tp + fp) if (tp + fp) > 0 else 0.0
    r = tp / (tp + fn) if (tp + fn) > 0 else 0.0
    f1 = (2 * p * r) / (p + r) if (p + r) > 0 else 0.0
    
    all_precisions.append(p)
    all_recalls.append(r)
    all_f1s.append(f1)

total_precision = sum(all_precisions) / len(all_precisions)
total_recall = sum(all_recalls) / len(all_recalls)
total_f1 = sum(all_f1s) / len(all_f1s)

print("Overall Results")
print(f"Total Precision: {total_precision*100:.1f}%")
print(f"Total Recall:    {total_recall*100:.1f}%")
print(f"Total F1 Score:  {total_f1*100:.1f}%")
```
